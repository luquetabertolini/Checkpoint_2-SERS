# Checkpoint 2 - SERS


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Algoritmo de regressão logística -----> Algoritmo de classificação
from sklearn.linear_model import LogisticRegression

# Separação dos dados de treino e teste
# X_train, y_train, X_test, y_test
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/luquetabertolini/Checkpoint_2-SERS/refs/heads/main/Data_for_UCI_named.csv")
df.head()
```

```python
# Inspeção inicial
df.shape
```

```python
df.info()
```

```python
# Atributos com valores numéricos
df.describe(include='number')
```

```python
# Atributos com valores categóricos
df.describe(include='object')
```

```python
# Renomeando as labels
df['stabf'] = df['stabf'].replace({'stable': 'REDE INSTÁVEL', 'unstable': 'REDE ESTÁVEL'})
```

```python
# Quais classes temos na coluna 'stabf' (target)?
df['stabf'].unique()
```

```python
# Quantos registros temos em cada classe?
df['stabf'].value_counts()
```

```python
# Criar uma variável chamada "y" com os dados da coluna 'stabf'

y = df['stabf']
print(y.head())
```

```python
# Criar uma variável "x" com os dados dos 12 atributos numéricos
x = df.drop('stabf', axis=1)
print(x.head())
```

```python
# Separando dados para treino e para teste
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
modelo = LogisticRegression()
modelo.fit(X_train, y_train)
```

```python
# Probabilidades das previsões
probabilidades = modelo.predict_proba(X_test)
probabilidades[:30]
```

```python
# Gerando array com as previsões
y_predict = modelo.predict(X_test)
y_predict[:30]
```

```python
# Método de avaliação
acuracia = accuracy_score(y_test, y_predict)
print(f'Acurácia {acuracia*100:.2f}%')
```
